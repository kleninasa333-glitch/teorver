### Задача 1.48
> В старинной игре в кости необходимо было для выигрыша получить при бросании трех игральных костей сумму очков, превосходящую 10. Найти вероятности: а) выпадения l l очков; б) выигрыша.

```python
arr = list(itertools.product(range(1, 7), repeat = 3))
n = len(arr)

sum_11 = [x for x in arr if sum(x) == 11]
round((len(sum_11) / n), 4)
```

## Задание 2.9
В среднем 20% пакетов акций на аукционах продаются по первоначально заявленной цене. Найти вероятность того, что из 9 пакетов акций в результате торгов по первоначально заявленной цене: 1) не будут проданы 5 пакетов; 2) будет продано: а) менее 2 пакетов; б) не более 2 пакетов; в) хотя бы 2 пакета; г) наивероятнейшее число пакетов.

Дано: n = 9, p = 0.2, q = 0.8

1. Не будут проданы 5 пакетов -> значит продано ровно 4 пакета (из 9). По формуле Бернулли: P_9(4) = C(9, 4) * 0.2^4 * 0.8^5

```python
round(stats.binom.pmf(4, 9, 0.2), 4)
```

2а. Менее 2 пакетов (0 или 1): P(m < 2) = P(0) + P(1)

```python
round(stats.binom.cdf(1, 9, 0.2), 4)
```

2б. Не более 2 пакетов (m <= 2): P(m <= 2) = P(0) + P(1) + P(2)

```python
round(stats.binom.cdf(2, 9, 0.2), 4)
```

2в. Хотя бы 2 пакета (от 2 до 9): P(m >= 2) = 1 - P(m < 2)

```python
round(1 - stats.binom.cdf(1, 9, 0.2), 4)
```

2г. Наивероятнейшее число пакетов m0: np - q <= m0 <= np + p 90.2 - 0.8 <= m0 <= 90.2 + 0.2 1 <= m0 <= 2 Получается m0 = 1 и m0 = 2 P = P_9(1) + P_9(2)

```python
round(stats.binom.pmf(1, 9, 0.2) + stats.binom.pmf(2, 9, 0.2), 4)
```

## Задание 2.11
По результатам проверок налоговыми инспекциями установлено, что в среднем каждое второе малое предприятие региона имеет нарушение финансовой дисциплины. Найти вероятность того, что из 1000 зарегистрированных в регионе малых предприятий имеют нарушения финансовой дисциплины: а) 480 предприятий; б) наивероятнейшее число предприятий; в) не менее 480; г) от 480 до 520.

n = 1000, p = 0.5, q = 0.5 np = 500, npq = 250, sqrt(npq) = sqrt(250) Используем теоремы Муавра-Лапласа

а) Ровно 480 предприятий (локальная теорема Муавра-Лапласа): x = (480 - 500) / sqrt(250) ~ -1.265 P_1000(480) ~ phi(x) / sqrt(npq)

```python
n = 1000
p = 0.5
s = math.sqrt(n * p * (1 - p))
x = (480 - n * p) / s
round(stats.norm.pdf(x) / s, 4)
```

б) Наивероятнейшее число предприятий: np - q <= m0 <= np + p => 500 - 0.5 <= m0 <= 500 + 0.5 => m0 = 500 x = (500 - 500) / sqrt(250) = 0

```python
x = (500 - n * p) / s
round(stats.norm.pdf(x) / s, 4)
```

в) Не менее 480 предприятий (от 480 до 1000) (интегральная теорема): P(480 <= m <= 1000) = Ф(x2) - Ф(x1)

```python
x1 = (480 - n * p) / s
x2 = (1000 - n * p) / s
round(stats.norm.cdf(x2) - stats.norm.cdf(x1), 4)
```

г) От 480 до 520 предприятий (интегральная теорема): P(480 <= m <= 520) = Ф(x2) - Ф(x1)

```python
x1 = (480 - n * p) / s
x2 = (520 - n * p) / s
round(stats.norm.cdf(x2) - stats.norm.cdf(x1), 4)
```
