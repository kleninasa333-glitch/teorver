### Задача 1.10
>Буквы Т, Е, И, Я, Р, О написаны на отдельных
карточках. Ребенок берет карточки в случайном порядке и прикладывает одну к друтой: а) 3 карточки; б) все 6 карточек. Какова вероятность того, что получится слово: а) «ТОР»; б) «ТЕОРИЯ»?

* a) $P(A) = \frac{1}{A_{6}^3} = \frac{3!}{6!}$

```python
round(sc.factorial(3) / sc.factorial(6),4)
```

```python
round(1/sc.perm(6,3),4)
```

* б) $P(A) = \frac{1}{P_{6}} = \frac{1}{6!}$

```python
round(1/sc.factorial(6),4)
```

### Задача 1.48
> В старинной игре в кости необходимо было для выигрыша получить при бросании трех игральных костей сумму очков, превосходящую 10. Найти вероятности: а) выпадения l l очков; б) выигрыша.
* a)$F(A) = \frac{27}{6^3}$

```python
arr = list(itertools.product(range(1, 7), repeat = 3))
n = len(arr)

sum_11 = [x for x in arr if sum(x) == 11]
round((len(sum_11) / n), 4)
```

* a)$F(A) = \frac{108}{6^3}$

```python
win = [x for x in arr if sum(x) > 10]
round((len(win) / n), 4)
```
